```python
def calculate_taxable_income(gross_income, pension, NHF, NHIS, Mortgage, Rent, LifeInsurance):
    """Calculate taxable income after deducting allowed expenses."""
    taxable_income = gross_income - pension - NHF - NHIS - Mortgage - Rent - LifeInsurance
    
    # Ensure taxable income is not negative
    if taxable_income < 0:
        taxable_income = 0
        
    return taxable_income


# Simple input version
def simple_input_calculator():
    """Simple calculator with one-time input."""
    print("\nEnter the following values:")
    gross_income = float(input("Gross Income: "))
    pension = float(input("Pension Contribution: "))
    NHF = float(input("NHF Contribution: "))
    NHIS = float(input("NHIS Contribution: "))
    Mortgage = float(input("Mortgage Interest: "))
    Rent = float(input("Rent Paid: "))
    LifeInsurance = float(input("Life Insurance Premium: "))
    
    # Calculate
    taxable_income = calculate_taxable_income(gross_income, pension, NHF, NHIS, Mortgage, Rent, LifeInsurance)
    total_deductions = pension + NHF + NHIS + Mortgage + Rent + LifeInsurance
    
    # Display results
    print(f"\nTaxable Income: ${taxable_income:,.2f}")
    print(f"Gross Income: ${gross_income:,.2f}")
    print(f"Total Deductions: ${total_deductions:,.2f}")
    print(f"Taxable Income: ${taxable_income:,.2f}")


# Run the simple version
simple_input_calculator()
```

```python
def calculate_tax(income):
    tax = 0

    if income <= 300000:
        tax = income * 0.07

    elif income <= 600000:
        tax = (300000 * 0.07) + ((income - 300000) * 0.11)

    elif income <= 1100000:
        tax = (300000 * 0.07) + (300000 * 0.11) + ((income - 600000) * 0.15)

    else:
        tax = (
            (300000 * 0.07)
            + (300000 * 0.11)
            + (500000 * 0.15)
            + ((income - 1100000) * 0.19)
        )

    return tax


# Get user input
income = float(input("Enter your annual income: "))

# Calculate tax
tax_amount = calculate_tax(income)

# Display result
print(f"Total tax payable: ₦{tax_amount:,.2f}")
```
