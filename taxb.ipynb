```python
# Python Program for Income Tax Calculation

def calculate_income_tax(taxable_income):
    tax_brackets = {
        0: 0.10,
        9875: 0.12,
        40125: 0.22,
        85525: 0.24,
        163300: 0.32,
        207350: 0.35,
        518400: 0.37
    }

    tax = 0

    for bracket, rate in tax_brackets.items():
        if taxable_income > bracket:
            taxable_amount = min(taxable_income - bracket, bracket - tax)
            tax += taxable_amount * rate
        else:
            break

    return tax

# Example usage
taxable_income = 50000
income_tax = calculate_income_tax(taxable_income)
print(f"The income tax for a taxable income of ${taxable_income} is ${income_tax}.")
```
